## The Viterbi Algorithm
Dynamic programming

$\pi_i$ = initial probabilities

* the chance that the first position is in state $i$

$a_{ij}$ = transition probabilities

* the chance that the next position is in state $j$ when this position is in state $i$

$b_j(k)$ = emission probabilities

* the chance that a position in state $j$ shows symbol $k$

Each row of the transition table and each row of the emission table adds up to 1, because each row lists every possible outcome from one starting point.

$\lambda = (A, B, \pi)$

* the three sets together

Assumptions (Markov property):

* The next state depends only on the current state, not on any earlier position
* Each observation depends only on the state at its own position

### Formal Description by Durbin et al.

Durbin writes emissions as $e_\ell(x_i)$ instead of $b_j(k)$; they mean the same thing. Durbin also adds a silent begin/end state, numbered 0.

#### Initialization ($i = 0$):

$$v_0(0) = 1, \quad v_k(0) = 0 \text{ for } k > 0$$

#### Recursion ($i = 1 \ldots L$):

$$v_\ell(i) = e_\ell(x_i) \cdot \max_k \big( v_k(i-1) \, a_{k\ell} \big)$$

$$\text{ptr}_i(\ell) = \arg\max_k \big( v_k(i-1) \, a_{k\ell} \big)$$

#### Termination:

$$P(x, \pi^*) = \max_k \big( v_k(L) \, a_{k0} \big)$$

$$\pi_L^* = \arg\max_k \big( v_k(L) \, a_{k0} \big)$$

#### Traceback ($i = L \ldots 1$):

$$\pi_{i-1}^* = \text{ptr}_i(\pi_i^*)$$

#### Glossary

* $\max$: the highest value
* $\arg\max$: which state gave the highest value (stored as the traceback pointer)
* $e_\ell(x_i)$: emission probability of symbol $x_i$ from state $\ell$
* $a_{k\ell}$: transition probability from state $k$ to state $\ell$
* $k$: the previous state (the one maximized over)
* $\ell$: the current state (the cell being filled)
* $L$: the length of the observation sequence
* $i$: the position in the sequence
* $x$: the observation sequence; $x_i$ is the symbol at position $i$
* $\pi^*$: the most probable path; $\pi_i^*$ is its state at position $i$ (not the same as the initial probabilities $\pi_i$)

### In our implementation

* Our model has no end state, so termination drops $a_{k0}$:

$$P(x, \pi^*) = \max_k v_k(L), \qquad \pi_L^* = \arg\max_k v_k(L)$$

* We are given initial probabilities directly, so column 1 is $v_k(1) = \pi_k \, e_k(x_1)$ and the recursion starts at $i = 2$.
* To avoid underflow, we work in log space (max stays max; products become sums):

$$\log v_\ell(i) = \log e_\ell(x_i) + \max_k \big( \log v_k(i-1) + \log a_{k\ell} \big)$$

#### Key Features
* Optimal path finding: identifies the most probable sequence of hidden states.
* Dynamic programming: uses efficient tabulation to avoid redundant calculations.
* Log-space computation: prevents numerical underflow with long sequences by turning products into sums
* * In log space, multiplication becomes addition, and the max over previous states must likewise be converted into a sum of their log probabilities
* Traceback mechanism: reconstructs the optimal state path after computation.

```python
import math


class HMMSetup:
    """ Stores the model, checks the inputs, and builds the log-space tables.
    Every algorithm mixin (Viterbi now, forward-backward later) relies on the
    attributes this class creates: states, alphabet, log_init, log_trans, log_emit."""

    def __init__(self, initial_probabilities, transition_probabilities, emission_probabilities): 
        """Store the model, derive states and alphabet from the dictionaries, 
        validate inputs, build log-space versions of tables.
        
        Args:
            initial_probabilities: dictionary mapping each state to its probability of being the first state
            transition_probabilities: nested dictionary where 
                transition_probabilities[from_state][to_state] is the
                probability of moving from from_state to to_state
            emission_probabilities: nested dictionary where
                emission_probabilities[state][symbol] is the probability
                that state emits that symbol
                
            
        Attributes set:
            states: list of state names, in the order of initial_probabilities
            alphabet: sorted list of every symbol in any emission row
            log_init, log_trans, log_emit: log-space copies of the three tables
                (zero probabilities become -inf)
                
        Raises:
            ValueError: if the dictionaries fail validation via validate()
        """
 
        ### pseudocode:
            # store 3 probability dictionaries as attribute
            # states = list of keys of initial_probabilities
            # alphabet = every symbol that appears as a key in any emission row
            # call validate() to check inputs and raise an error if something is wrong
            # call to_log_space() to build log versions of the 3 tables

        #keep original probabilities for reference 
        self.initial_probabilities = initial_probabilities
        self.transition_probabilities = transition_probabilities
        self.emission_probabilities = emission_probabilities

        #states = list of keys of initial_probabilities
        #come from dictionaries so they align with probabilities
        self.states = list(initial_probabilities.keys())

        #alphabet = every symbol that is a key in any emission row
        symbols = set() #no repeats
        for state in emission_probabilities:
            symbols.update(emission_probabilities[state])
        self.alphabet = sorted(symbols) #sorted so order is the same every time

        #check inputs before so log conversion has valid input and build log versions of tables
        self.validate()
        self.to_log_space()

        
        

    def validate(self): 
        """Check that probability dictioaries describe a valid HMM
        
        Checks:
            1. Keys: transition dict (rows and columns) and emission dict (rows) use the same states as initial_probabilities
            2. Range: every probability is between 0 and 1
            3. Sums: initial probabilities, each transition row, and each emission row sums to 1 (within a tolerance of 1e-6)

        Raiss:
            ValueError: naming dictionary and state that failed
        """

        ### pseudocode:
            # check that and raise a ValueError with a message showing what dict and state failed
            #   transition dict has the same state as initial (rows and columns)
            #   emission dict has the same states as initial (rows)
            #   initial probabilities sum to ~1
            #   each transition row sums to ~1
            #   each emission row sums to ~1
            #   every probability is between 0 and 1
        

        #1. keys, every table must have the same set of states
        expected = set(self.states)
        transition_keys = set(self.transition_probabilities.keys())
        if transition_keys != expected:
            raise ValueError(f"Transition states {transition_keys} do not match initial states {expected}")
        for from_state, row in self.transition_probabilities.items():
            to_keys = set(row.keys())
            if to_keys != expected:
                raise ValueError(f"Transition row for {from_state} has states {to_keys}, expected {expected}")
        emission_keys = set(self.emission_probabilities.keys())
        if emission_keys != expected:
            raise ValueError(f"Emission states {emission_keys} do not match initial states, {expected}")
        
        #emission rows must all list the same symbols, a state missing a symbol would pass this check but cause a KeyError later in the recursion
        expected_symbols = set(self.alphabet)
        for state, row in self.emission_probabilities.items():
            symbol_keys = set(row.keys())
            if symbol_keys != expected_symbols:
                raise ValueError(f"Emission row for {state} has symbols {symbol_keys}, expected {expected_symbols}")
        
        #2. range, a row can sum to 1 and still have an invalid value so each probability is checked
        for init_state, prob in self.initial_probabilities.items():
            if prob < 0 or prob > 1:
                raise ValueError(f"Initial probability {prob} for state {init_state} must be between 0 and 1.")
        for from_state, row in self.transition_probabilities.items():
            for to_state, prob in row.items():
                if prob < 0 or prob > 1:
                    raise ValueError(f"Transition probability {prob} from {from_state} to {to_state} must be between 0 and 1")
        for state, row in self.emission_probabilities.items():
            for symbol, prob in row.items():
                if prob < 0 or prob > 1:
                    raise ValueError(f"Emission probability {prob} of {symbol} in state {state} must be between 0 and 1")
        
        #3. sums, each set of options must sum to 1, a tolerance is used because floating point numbers rarely add to exactly 1
        tolerance = 1e-6
        init_total = sum(self.initial_probabilities.values())
        if abs(init_total - 1) > tolerance:
            raise ValueError(f"Initial probabilities sum to {init_total}, not 1")
        
        for from_state, row in self.transition_probabilities.items():
            total = sum(row.values())
            if abs(total - 1) > tolerance:
                raise ValueError(f"Transition row for {from_state} sums to {total}, not 1")
        
        for state, row in self.emission_probabilities.items():
            total = sum(row.values())
            if abs(total - 1) > tolerance:
                raise ValueError(f"Emission row for {state} sums to {total}, not 1")

    def safe_log(self, p):
        """Return the natural log of p, or -inf if p is 0"""
        #if p is 0, return negative infinity, otherwise return the natural log of p
        if p == 0:
            return float('-inf')
        return math.log(p)
        
    def to_log_space(self): # (convert to log space and handle zeros)
        """ Build log-space copies of the probability tables
        
        Viterbi multiplies many probabilities below 1, which underflows to 0 for long sequences. 
        In log space, products became sums, which stay in range. Zero probabilities become -inf
        
        Attributes set:
            log_init: dict, log_init[state]
            log_trans: nested dict, log_trans[from_state][to_state]
            log_emit: nested dict, log_emit[state][symbol]

        Original dicts are kept unchanged
        """
        ### pseudocode:
        #   build log versions of all 3 tables (self.log_init, self.log_trans, self.log_emit)
        #   handle log 0 = -inf
        #       if p = 0, store float('-inf') since a zero probability move is impossible

        #initial probabilites
        self.log_init = {}
        for state, prob in self.initial_probabilities.items():
            self.log_init[state] = self.safe_log(prob)
        self.log_trans = {}

        #transitions: nested dict, each row is built separatey and stored under from-state
        for from_state, row in self.transition_probabilities.items():
            log_row = {}
            for to_state, prob in row.items():
                log_row[to_state] = self.safe_log(prob)
            self.log_trans[from_state] = log_row

        #emissions: same nested pattern, symbols are inner keys    
        self.log_emit = {}
        for state, row in self.emission_probabilities.items():
            log_row = {}
            for symbol, prob in row.items():
                log_row[symbol] = self.safe_log(prob)
            self.log_emit[state] = log_row

```

```python
class ViterbiMixin:
    """
    Adds Viterbi decoding (most probable hidden-state path).
    This can't be used on its own, because it uses self.states, self.log_init,
    self.log_trans and self.log_emit, which HMMSetup creates.
    Always use it through HiddenMarkovModel.
    """        
    def initialization_viterbi(self, sequence): # for each state, log initial + log emission of the first symbol
        """ Create the score and arrow grids and fill the first position
        
        The first position has no previous state, so its score is
        log initial probability + log emission of the first symbol:
            
            score[0][state] = log_init[state] + log_emit[state][sequence[0]]
            
            positions 1 onward are filled by calculate_viterbi_score.
            
            Args: 
                sequence: the observed symbols (string or list)
            
            Attributes set:
                score: list of dictionaries; score[i][state] is the log probability of
                    the best path that covers positions 0..1 and ends in state.
                arrows: list of dictionaries: arrows[i][state] is the state at position
                    i-1 on that best path"""
        ### pseudocode:
        #   create score grid: self.score[i][state] -> log score at position i, state
        #   create arrow grid: self.arrows[i][state] -> which previous state won the max
        #   fill only the first position (index 0, symbol = sequence[0]):
        #   for each state:
        #       score = log initial[state] + log emission[state][sequence[0]]
        #       arrow = None (no previous state yet)
        #   positions after are filled by calculate_viterbi_score
        
        #one dictionary per position, appended in order so score[i] is position i
        self.score = []
        self.arrows = []

        score_column = {}
        arrow_column = {}
        for state in self.states:
            #log space so initial x emission become initial + emission
            score_column[state]= self.log_init[state] + self.log_emit[state][sequence[0]]
            #no previous state because we're at position 0
            arrow_column[state] = None

        self.score.append(score_column)
        self.arrows.append(arrow_column)


    def calculate_viterbi_score(self, sequence): # recursion and nested loops (positions x current state x previous state)
        """Fill the score and arrow grids for positions 1 onward
        
        For each position i and each current state, tries every possible state and keeps the best:
            
            score[i][current] = log_emit[current][sequence[i]] +
                                max over previous of (score[i-1][previous] + log_trans[previous][current])
                                
        winning previous state is stored in arrows[i][current] for traceback
        
        assumes initialization_viterbi has already filled position 0
        
        Args:
            sequence: the observed symbols, the same sequence passed to initialization_viterbi"""
        #Pseudocode
        #for each position i from 1 to len(sequence) - 1:
        #   symbol = sequence[i]
        #   score_column = empty dict
        #   arrow_column = empty dict
        #   for each current state:
        #       best_value = -inf
        #       best_previous = None
        #       for each previous state:
        #           candidate = self.score[i-1][previous] + log transition[previous][current]
        #           if candidate > best_value:
        #               best_value = candidate
        #               best_previous = previous
        #       score_column[current] = best_value + log emit[current][symbol]
        #       arrow_column[current] = best_previous
        #   append score column to self.score
        #   append arrow column to self.arrows

        for i in range(1, len(sequence)):
            symbol = sequence[i]
            score_column = {}
            arrow_column = {}

            for current_state in self.states:
                #start below real scores so first candidate wins
                best_value = float('-inf')
                best_previous = None

                #try every state we could have come from
                for previous_state in self.states:
                    candidate = self.score[i-1][previous_state] + self.log_trans[previous_state][current_state]
                    if candidate > best_value:
                        best_value = candidate
                        best_previous = previous_state
                
                #emission is same for eevery previous state so it is added once after the max
                score_column[current_state] = best_value + self.log_emit[current_state][symbol]
                arrow_column[current_state] = best_previous

            self.score.append(score_column)
            self.arrows.append(arrow_column)


    def termination_viterbi(self):
        """ Find the most probable state and the path's log probability
        
        Picks the state with the highest score in the last column
        
            best_score = max over states of score[L-1][state]
            
        Returns:
            (best_state, best_score): the last state of the most probable path and that path's log probability
            
        Raises: 
            ValueError: if every final score is -inf there is no possible path"""
    #pseudocode
    #   best_score = -inf
    #   best_state = none
    #   for state in last column of self.score (self.score[-1]):
    #       if self.score[-1][state] > best_score:
    #           best_score = self.score[-1][state]
    #           best_state = state
    #   if best_state is None, raise error (every path impossible)
    #   return best_state, best_score
        last_column = self.score[-1]

        #start below real scores so a state can only be chosen if path is possible
        best_score = float('-inf')
        best_state = None
        for state in last_column:
            if last_column[state] > best_score:
                best_score = last_column[state]
                best_state = state
        
        #best_state stays None only if every score was -inf
        if best_state is None:
            raise ValueError("Every path has probability 0; this model cannot generate the sequence")
        return best_state, best_score    
 
    def traceback_viterbi(self, final_state): 
        """Follow the arrows back from the final state to get the path
        
        arrows[i][state] gives the state at position i-1, so starting from
        the last position and stepping back one arrow at a time rebuilds the path
        from the end to the start. loop stops at position 1
        
        Args: 
            final_state: the last state of the best path, comes from termination_viterbi
            
        Returns:
            list of states, first position to last
        """

    # path = [final_state]
    # current = final_state
    # for i from len(self.arrows) - 1 backwards to 1:
    #   current = self.arrows[i][current] #state at position i-1
    #   append current to path
    #   reverse path
    #   return path

        path = [final_state]
        current = final_state

        #range's stop value (0) is excluded
        for i in range((len(self.arrows)-1), 0, -1):
            current = self.arrows[i][current]
            path.append(current)
        
        #path was built from last-to-first so reverse it 
        path.reverse()
        return path
    

    def viterbi(self, sequence):
        """Find the most probable hidden-state path for an observed sequence
        
        Checks the input, runs the 4-steps of the Viterbi algorithm:
        initialization, recursion, termiantion, and traceback
        
        Args:
            sequence: the observed sequence (string or list)
        
        Returns:
                dict with keys:
                "path": the most probable state path as a list, first position to last
                "log_probability": its natural-log joint probability with the sequence
                "score": the score table, a list with one {state: log score} dict per position
                "arrows": the traceback table, a list with one {state: previous state} dict per position
            
        Raises:
            ValueError: if the sequence is empty, or contains a symbol that no state can emit"""
    #pseudocode:
    # if sequence is empty, raise ValueError
    # for each position and symbol in sequence:
    #   if no state has nonzero emission probability for a symbol
    #       raise ValueError that names the symbol and position
    # initialization_viterbi(sequence) #fills position 0
    # calculate_viterbi_score(sequence) #fills position 1 onward
    # best_state, best_score = termination_viterbi()
    # path = traceback_viterbi(best_state)
    # return {"path", "log_probability", "score", "arrows"}
    
        if len(sequence) == 0:
            raise ValueError("The sequence is empty")
        
        #catch impossible symbols up front
        for position, symbol in enumerate(sequence):
            can_emit = False
            for state in self.states:
                if self.emission_probabilities[state].get(symbol,0) > 0:
                    can_emit = True
                    break
            if not can_emit:
                raise ValueError(f"No state can emit symbol {symbol} at position {position}.")
        
        self.initialization_viterbi(sequence) #position 0
        self.calculate_viterbi_score(sequence) #positions 1 onward
        best_state, best_score = self.termination_viterbi()
        path = self.traceback_viterbi(best_state)

        #return the tables too, so later projects can reuse them
        return {"path": path,
                "log_probability": best_score,
                "score": self.score,
                "arrows": self.arrows}
```

```python
class HiddenMarkovModel(HMMSetup, ViterbiMixin):
    """
    A Hidden Markov model built from initial, transition, and emission probability dictionaries
    
    Implements Viterbi decoding (finding the most probable hidden-state path).
    Created so later algorithms can reuse the same stored model and tables
    """
init_probs = {"I": 0.2, "G": 0.8}
trans_probs = {"I": {"I": 0.7, "G": 0.3},
               "G": {"I": 0.1, "G": 0.9}}
emit_probs = {"I": {"A": 0.1, "C": 0.4, "G": 0.4, "T": 0.1},
              "G": {"A": 0.3, "C": 0.2, "G": 0.2, "T": 0.3}}
hmm = HiddenMarkovModel(init_probs, trans_probs, emit_probs)

hmm.initialization_viterbi("G")
print(hmm.score)
print(hmm.arrows)

hmm.initialization_viterbi("GG")
hmm.calculate_viterbi_score("GG")
print(hmm.score)
print(hmm.arrows)

c = HiddenMarkovModel({"F": 0.5, "L": 0.5},
                      {"F": {"F": 0.6, "L": 0.4}, "L": {"F": 0.4, "L": 0.6}},
                      {"F": {"H": 0.5, "T": 0.5}, "L": {"H": 0.8, "T": 0.2}})
c.initialization_viterbi("HHTT")
c.calculate_viterbi_score("HHTT")
print(c.termination_viterbi())

best_state, best_score = c.termination_viterbi()
print(c.traceback_viterbi(best_state))


bad_trans = {"I": {"I": 0.7, "X": 0.3},
             "G": {"I": 0.1, "G": 0.9}}
try:
    HiddenMarkovModel(init_probs, bad_trans, emit_probs)
except ValueError as e:
    print("Caught:", e)

bad_sum = {"I": {"I": 0.7, "G": 0.2},
           "G": {"I": 0.1, "G": 0.9}}
try:
    HiddenMarkovModel(init_probs, bad_sum, emit_probs)
except ValueError as e:
    print("Caught:", e)

bad_emit = {"I": {"A": -0.1, "C": 0.6, "G": 0.4, "T": 0.1},
            "G": {"A": 0.3, "C": 0.2, "G": 0.2, "T": 0.3}}
try:
    HiddenMarkovModel(init_probs, trans_probs, bad_emit)
except ValueError as e:
    print("Caught:", e)

uneven_emit = {"I": {"A": 0.5, "C": 0.5},
               "G": {"A": 0.3, "C": 0.2, "G": 0.2, "T": 0.3}}
try:
    HiddenMarkovModel(init_probs, trans_probs, uneven_emit)
except ValueError as e:
    print("Caught:", e)
```
